```python
import matplotlib.pyplot as plt
import numpy as np

plt.rcParams.update({
    "text.usetex": True,
})
```

```python
vsa_latency = 4
fast_bhm_latency = 460
bhm_latency = 22350

num_epochs = 100

vsa_latencies = np.arange(0, num_epochs, 1) * vsa_latency
fast_bhm_latencies = np.arange(0, num_epochs, 1) * fast_bhm_latency
bhm_latencies = np.arange(0, num_epochs, 1) * bhm_latency
```

```python
fig = plt.figure(figsize=(5, 3))
plt.plot(vsa_latencies, label='VSA', color='green', linestyle='--')
plt.plot(fast_bhm_latencies, label='Fast BHM', color='orange', linestyle='--')
plt.plot(bhm_latencies, label='BHM', color='blue', linestyle='--')
plt.title('Cumulative Latency per Scan')
plt.xlabel('Number of Scans')
plt.ylabel('Latency (ms)')
plt.yscale('log')
plt.legend()
```
