```python
import pandas as pd 
import wandb
import json

api = wandb.Api()

# Project is specified by <entity/project-name>
runs = api.runs("ssnyde9-george-mason-university/decoding_method_evaluation_ablation_sparse_no_entropy")

count = 0

config_list = []
for run in runs: 
    if count % 10 == 0:
        print(f"Run {count} of {len(runs)}")
    count += 1

    run_config = {}
    run_config["test_nll"] = run.summary["test_nll"]
    run_config["test_auc"] = run.summary["test_auc"]

    # .config contains the hyperparameters.
    #  We remove special values that start with _.
    mapping_config = run.config["mapping"]
    for key, value in mapping_config.items():
            if isinstance(value, dict):
                for k, v in value.items():
                    run_config[f"{key}_{k}"] = v
            else:
                run_config[key] = value

    config_list.append(run_config)

runs_df = pd.DataFrame(config_list)

runs_df.to_csv("exp1_ablation_sparse_no_entropy.csv")
```
